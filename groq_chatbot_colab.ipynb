## Step 3: Set up the Groq client

```python
from groq import Groq

client = Groq(api_key=api_key)

MODEL = "llama-3.3-70b-versatile"   # you can change to another Groq-hosted model
SYSTEM_PROMPT = "You are a helpful, friendly assistant."

# Conversation memory
messages = [{"role": "system", "content": SYSTEM_PROMPT}]

print("Client ready. Model:", MODEL)
```

## Step 4: Helper function to chat

Run this cell once to define the function, then use the next cell to actually chat.

```python
def chat(user_message):
    messages.append({"role": "user", "content": user_message})

    response = client.chat.completions.create(
        model=MODEL,
        messages=messages,
        temperature=0.7,
        max_tokens=1024,
    )

    reply = response.choices[0].message.content
    messages.append({"role": "assistant", "content": reply})

    print("You:", user_message)
    print("Bot:", reply)
    return reply
```

## Step 5: Chat with the bot

Run this cell as many times as you like, changing the message each time.
Since `messages` is stored outside the function, the bot remembers earlier turns.

```python
chat("Hi! Who are you?")
```

```python
chat("Can you help me write a poem about the ocean?")
```

## Optional: Interactive loop (like a terminal chat)

Run this cell to get a continuous chat inside Colab. Type `exit` to stop.
Note: this uses `input()`, which works fine in Colab but pauses the notebook until you respond.

```python
while True:
    user_input = input("You: ")
    if user_input.lower() in ("exit", "quit"):
        print("Bot: Goodbye!")
        break
    chat(user_input)
```
