```python
# Personal Information

name = input("Enter Your Name : ")

email = input("Enter Email : ")

phone = input("Enter Phone Number : ")

address = input("Enter Address : ")
```

```python
objective = input("Enter Career Objective : ")
```

```python
education = input("Enter Education : ")
```

```python
skills = input("Enter Skills (comma separated): ")
```

```python
projects = input("Enter Projects (comma separated): ")
```

```python
certifications = input("Enter Certifications : ")
```

```python
languages = input("Enter Languages Known : ")
```

```python
resume = f"""

==================================================
                 PROFESSIONAL RESUME
==================================================

Name          : {name}

Email         : {email}

Phone         : {phone}

Address       : {address}


Career Objective
--------------------------------------------------

{objective}


Education
--------------------------------------------------

{education}


Skills
--------------------------------------------------

{skills}


Projects
--------------------------------------------------

{projects}


Certifications
--------------------------------------------------

{certifications}


Languages
--------------------------------------------------

{languages}

==================================================

"""
```

```python
print(resume)
```

```python
with open("Resume.txt","w") as file:

    file.write(resume)

print("Resume Saved Successfully")
```
