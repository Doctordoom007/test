```python
# Chatbot Knowledge Base

knowledge = {

    # Greetings
    "hello":"Hello! Welcome to the AI Chatbot.",
    "hi":"Hi! How can I help you?",
    "good morning":"Good Morning!",
    "good afternoon":"Good Afternoon!",
    "good evening":"Good Evening!",

    # Introduction
    "who are you":"I am a Rule-Based AI Chatbot built using Python.",
    "your name":"My name is AI Chatbot.",

    # AI
    "what is ai":"Artificial Intelligence is the simulation of human intelligence by machines.",

    # ML
    "what is machine learning":"Machine Learning is a subset of AI that learns patterns from data.",

    # DL
    "what is deep learning":"Deep Learning is a subset of Machine Learning based on neural networks.",

    # NLP
    "what is nlp":"Natural Language Processing enables computers to understand human language.",

    # Transformer
    "what is transformer":"Transformer is a deep learning architecture mainly used for NLP tasks.",

    # GPT
    "what is gpt":"GPT stands for Generative Pre-trained Transformer.",

    # BERT
    "what is bert":"BERT stands for Bidirectional Encoder Representations from Transformers.",

    # Sentiment Analysis
    "what is sentiment analysis":"Sentiment Analysis predicts whether a text expresses positive, negative or neutral sentiment.",

    # Spam Detection
    "what is spam detection":"Spam Detection identifies whether a message is spam or ham.",

    # Text Summarization
    "what is text summarization":"Text Summarization converts a long article into a short meaningful summary.",

    # Python
    "what is python":"Python is a popular programming language used in AI, Machine Learning and Data Science.",

    # Data Science
    "what is data science":"Data Science is the process of extracting useful insights from data.",

    # Thanks
    "thank you":"You're Welcome!",

    "thanks":"My pleasure!",

    # Bye
    "bye":"Goodbye! Have a great day!"
}
```

```python
# Start Chatbot

while True:

    user = input("You : ").lower().strip()

    # Exit
    if user == "exit":
        print("Bot : Chatbot Closed Successfully.")
        break

    # Exact Match
    elif user in knowledge:
        print("Bot :", knowledge[user])

    # Unknown Question
    else:
        print("Bot : Sorry! I don't know the answer to this question.")
```
